```python
import matplotlib.pyplot as plt
import random

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']


x = [i for i in range(25)]
y_bj = [random.randint(0,15) for i in range(25)]
y_xa = [random.randint(13,18) for i in range(25)]

plt.figure(figsize=(10,5))#设置画布大小

plt.plot(x,y_bj, label='北京') #折线图 如果没有画布，会自动创建一个
plt.plot(x,y_xa, label='西安') #折线图 如果没有画布，会自动创建一个

#设置折线图详细信息
plt.title('气温变化折线图')
plt.xlabel('时间')
plt.ylabel('温度')
plt.xticks(x)
y_ticks = [i for i in range(0 ,21)]
plt.yticks(y_ticks)
plt.grid()
plt.legend(loc = 'best')


plt.show()
```
