```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize
from sympy.abc import alpha

plt.rcParams['font.sans-serif'] = ['SimHei']

figure,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,6),dpi=500)

countries = ['china','english','japan','korea','america','russia']
values = [35,23,18,21,30,25]

axes1 :Axes = axes[0]
axes1.bar(countries,values,width = 0.6,color = 'r')
axes1.set_xlabel('国家',fontsize=14)

axes1.set_ylabel('石油储备(亿吨)',fontsize=14)
axes1.grid(linestyle='--',alpha=0.5)

axes1.set_title('世界石油储备',fontsize=18)

#饼状图
countries2 = ['India','China','America','Indonesia','Pakistan','Nigeria']
values= [14.51,14.09,3.4,2.83,2.51,2.33]

axes2 :Axes = axes[1]
axes2.pie(values,labels=countries2,autopct='%1.1f%%',)
axes2.set_title('世界人口比例',fontsize =18)
axes2.legend(loc = 'best')

#保存文件
plt.savefig('world_oil_reserves.png')
plt.show()
```
